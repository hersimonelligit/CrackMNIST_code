```python
# ==========================================
# INSPECCIÓN DEL DATASET DE SIF REGRESSION
# ==========================================

import numpy as np
import matplotlib.pyplot as plt

sif_dataset = CrackMNIST(
    split="train",
    pixels=28,
    size="S",
    task="SIF_regression"
)

print(f"Samples: {len(sif_dataset)}")
print(f"Description:\n{sif_dataset.description}")

# Primer sample
x, y = sif_dataset[0]

print("\n--- INPUT ---")
print("Type:", type(x))
print("Shape:", x.shape)
print("Dtype:", x.dtype)

print("\n--- TARGET ---")
print("Type:", type(y))
print("Value:", y)

try:
    print("Shape:", y.shape)
except:
    print("Shape:", np.array(y).shape)

# Inspeccionar varios targets
print("\n--- FIRST 10 TARGETS ---")

targets = []

for i in range(10):
    _, target = sif_dataset[i]
    targets.append(target)
    print(f"{i}: {target}")
```

```python
targets = np.array([
    sif_dataset[i][1]
    for i in range(len(sif_dataset))
])

print("Targets shape:", targets.shape)

print("\nMIN:")
print(targets.min(axis=0))

print("\nMAX:")
print(targets.max(axis=0))

print("\nMEAN:")
print(targets.mean(axis=0))

print("\nSTD:")
print(targets.std(axis=0))
```

```python
from crackmnist import CrackMNIST

train_dataset = CrackMNIST(
    split="train",
    pixels=28,
    size="S",
    task="SIF_regression"
)

val_dataset = CrackMNIST(
    split="validation",
    pixels=28,
    size="S",
    task="SIF_regression"
)

test_dataset = CrackMNIST(
    split="test",
    pixels=28,
    size="S",
    task="SIF_regression"
)

print("Train:", len(train_dataset))
print("Validation:", len(val_dataset))
print("Test:", len(test_dataset))
```
